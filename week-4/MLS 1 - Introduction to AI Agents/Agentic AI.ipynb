# Type os Agents

##1. Reactive Agents

A **Reactive Agent** responds directly to current inputs.
It does not plan ahead, maintain long-term goals, or build an internal model of the world.

It follows:

    Perception → Action

No memory of past states.
No future prediction.

**Characteristics**

* Rule-based or policy-based

* Fast decision making

* No internal planning

* Stateless (or minimal state)

* Trigger-response behavior

**Examples**

**Traffic Light Controller**

* Detects cars → changes signal

* Does not “plan” traffic strategy for the day

* Reacts to sensor input only


**Robotic Vacuum (Basic Model)**

* Hits wall → turns

* Detects dirt → cleans

* No global map of the house

**Basic Chatbot**

* User asks question

* Bot answers

* No memory of past conversations



**In AI/ML Terms**

* Reactive agents are common in:

* Rule engines

* Simple control systems

* Basic reinforcement learning policies (no planning layer)

##2. Proactive Agents

A **Proactive** Agent is goal-driven.
It takes initiative and acts toward a future objective — even without immediate triggers.

It follows:

    Goal → Plan → Act → Monitor → Adjust

**Characteristics**

* Has internal goals

* Performs planning

* Maintains internal state

* Anticipates future conditions

* Takes initiative

**Examples**

**Smart Calendar Assistant**

* Notices schedule conflict

* Suggests alternative time

* Sends reminders before meeting

* It doesn’t wait for you to ask.

**Autonomous Vehicle**

* Predicts pedestrian movement

* Adjusts speed in advance

* Plans optimal route

* It anticipates.

**Investment Portfolio Rebalancer**

* Detects drift in allocation

* Automatically adjusts portfolio

* Acts before losses escalate

**In Agentic AI Context**

A proactive LLM agent might:

* Monitor project deadlines

* Retrieve relevant policies

* Send compliance alerts

* Suggest mitigation plans

It doesn’t wait for explicit instructions.

##3. Interactive Agents

An **Interactive Agent** engages in continuous communication with users or other agents.
It adapts behavior based on dialogue and feedback.

It follows:

    Observe → Communicate → Update → Act → Continue

**Characteristics**

* Two-way communication

* Context-aware

* Learns from feedback

* Adjusts responses dynamically

* Often social or collaborative

**Examples**

**Customer Support AI**

* Asks clarifying questions

* Adjusts troubleshooting steps

* Escalates when needed

**AI Tutor**

* Evaluates student answer

* Adjusts difficulty

* Provides hints

**Multi-Agent Negotiation System**

* Agents exchange offers

* Adapt strategies

* Reach consensus

## Comparison Table

Feature|	Reactive|	Proactive|	Interactive
--|--|--|--
Memory|	Minimal|	Yes|	Yes
Planning|	No|	Yes|	Sometimes
Goal-driven|	No|	Yes|	Yes
Initiative|	No|	Yes|	Yes
Dialogue-based|	No|	Not necessarily|	Yes
Adaptation|	Limited|	Moderate|	High

##Where Do LLM Agents Fit?

**Most modern LLM agents are:**

* Interactive (conversation-based)

* Proactive (goal-driven loops)

* Sometimes reactive (tool-triggered)

**Example:**

* An RFP Analysis Agent:

      Reactive → extracts requirements when asked

      Proactive → checks compliance gaps automatically

      Interactive → asks user clarifying questions

**Good to know**

* They represent increasing sophistication:

      Reactive → Proactive → Interactive → Autonomous Multi-Agent Systems

**Also remember**

* A reactive system responds.
* A proactive system plans.
* An interactive system collaborates.

#Multi-Agent Systems (MAS)

##What Is a Multi-Agent System (MAS)?

A system composed of multiple autonomous agents that interact within a shared environment to achieve individual or collective goals.

Instead of:

    One Agent → One Goal


We have:

    Agent A ↔ Agent B ↔ Agent C → Shared Environment


Each agent:

* Has its own perception

* Has its own policy

* May have its own objective

* Interacts with others

##Why Do We Need Multi-Agent Systems?

Single agents struggle when:

* The environment is too complex

* Tasks require specialization

* Goals conflict

* Scale is large

* Coordination is required

Multi-agent systems allow:

* Division of labor

* Parallelism

* Negotiation

* Cooperation or competition

##What are the Core Characteristics of MAS?

1. Autonomy – Each agent makes its own decisions

2. Local perception – No agent sees the whole environment

3. Interaction – Communication or indirect coordination

4. Emergent behavior – Global behavior arises from local actions

##What are the Types of Multi-Agent Interactions?

**1. Cooperative**

Agents share a common goal.

Example:

* Search-and-rescue drones coordinating to cover different zones.

**2. Competitive**

Agents have opposing goals.

Example:

* Stock market trading bots.

* Chess AI vs opponent.

**3. Mixed (Co-opetition)**

Agents cooperate locally but compete globally.

Example:

* AI Delivery Drones from Competing Companies Sharing Airspace

LangGraph = a graph-based controller for agent workflows:

* Nodes = agents (or tools, evaluators, routers)

* Edges = who hands work to whom (the flow)

* State = shared memory passed between nodes (messages, plan, scratch, results)

##What "state" means?

Think of state like a shared folder that all nodes can read/write:

* messages (conversation so far)

* task (goal)

* plan

* draft

* reviews

* final_answer

##Pattern A — Centralized “Coordinator + Specialists”

When to use

* You want one brain to orchestrate work

* Clear accountability + easy governance

* Great for: “Research → Draft → Review” pipelines

image.png

```python
from typing import TypedDict, List
from langgraph.graph import StateGraph, END

class State(TypedDict):
    task: str
    research: str
    draft: str
    critique: str
    final: str
    _next: str # Add _next to the State TypedDict

def coordinator(state: State) -> State:
    # Decide which specialist to call next based on missing fields
    if not state.get("research"):
        state["_next"] = "research_agent"
    elif not state.get("draft"):
        state["_next"] = "writer_agent"
    elif not state.get("critique"):
        state["_next"] = "critic_agent"
    else:
        state["_next"] = "finalize"
    return state

def research_agent(state: State) -> State:
    state["research"] = f"Find key facts for: {state['task']}"
    return state

def writer_agent(state: State) -> State:
    state["draft"] = f"Draft based on research: {state['research']}"
    return state

def critic_agent(state: State) -> State:
    state["critique"] = "Critique: improve clarity, add structure."
    return state

def finalize(state: State) -> State:
    state["final"] = f"{state['draft']}\n\nEdits: {state['critique']}"
    return state

g = StateGraph(State)
g.add_node("coordinator", coordinator)
g.add_node("research_agent", research_agent)
g.add_node("writer_agent", writer_agent)
g.add_node("critic_agent", critic_agent)
g.add_node("finalize", finalize)

g.set_entry_point("coordinator")

# Conditional routing based on coordinator decision
def route(state: State):
    return state["_next"]

g.add_conditional_edges("coordinator", route, {
    "research_agent": "research_agent",
    "writer_agent": "writer_agent",
    "critic_agent": "critic_agent",
    "finalize": "finalize",
})

# After each specialist, go back to coordinator
g.add_edge("research_agent", "coordinator")
g.add_edge("writer_agent", "coordinator")
g.add_edge("critic_agent", "coordinator")
g.add_edge("finalize", END)

app = g.compile()

'''
# Option A) Print Mermaid diagram text (copy/paste into https://mermaid.live)
print("\n--- Mermaid diagram ---")
print(app.get_graph().draw_mermaid())
'''

'''
# Option B) Save a PNG image of the graph (requires LangGraph's mermaid rendering support)
png_bytes = app.get_graph().draw_mermaid_png()
with open("agentic_ai_langgraph.png", "wb") as f:
    f.write(png_bytes)
print("\nSaved graph image to: agentic_ai_langgraph.png")
'''

# =========================
# RUN THE GRAPH
# =========================
result = app.invoke({"task": "Create a 1-page intro to Agentic AI"})
print(result["final"])
```

###What are the differences between Centralized and Sequential structures?

Dimension|	Sequential|	Coordinator-Centric
--|--|--
Flow|	Fixed|	Dynamic
Order|	Predefined|	Decided at runtime
Loops|	No|	Yes
Adaptation|	None|	Yes
Intelligence Level|	Low|	Higher
Agent Autonomy|	Minimal|	Structured autonomy
Failure Recovery|	Hard|	Possible


##Pattern B — Hierarchical MAS (Planner → Executors)

This is “management + workers”. Very effective for complex tasks.

When to use

* You need a plan and multiple subtasks

* Great for: report generation, solution designs, proposal writing

Example idea

* Task: “Write a project charter”

* Planner: creates sections + assigns

* Executors: write sections

* Aggregator: merges + formats

LangGraph tip: Put plan: List[str] and sections: dict into State.

image.png

```python
"""
LangGraph Demo: Pattern B — Hierarchical MAS (Planner → Executors → Aggregator)

What this demo shows (no LLM required):
- Planner breaks a task into a list of subtasks (a "plan")
- Executors run each subtask one-by-one (could be different specialist agents)
- Aggregator combines all executor outputs into a final deliverable

Install (if needed):
    pip install langgraph

Run:
    python hierarchical_mas_demo.py
"""

from __future__ import annotations
from typing import TypedDict, List, Dict, Literal, Optional
from langgraph.graph import StateGraph, END


# -----------------------------
# 1) Define shared state (the "blackboard")
# -----------------------------
class State(TypedDict, total=False):
    task: str
    plan: List[str]                 # list of subtasks from planner
    current_step: int               # which subtask we are executing
    chosen_executor: Literal["exec_research", "exec_write", "exec_risk", "exec_schedule"]
    results: Dict[str, str]         # subtask -> output text
    final: str                      # aggregated output


# -----------------------------
# 2) Planner node
# -----------------------------
def planner(state: State) -> State:
    task = state["task"]

    # Simple heuristic plan generator (replace with LLM planner later)
    plan = [
        f"Research key facts about: {task}",
        f"Draft a concise response for: {task}",
        f"Identify risks/limitations for: {task}",
        f"Create a short schedule/checklist to deliver: {task}",
    ]

    return {
        "plan": plan,
        "current_step": 0,
        "results": {},
    }


# -----------------------------
# 3) Router: choose which executor to use for the current plan step
# -----------------------------
def route_to_executor(state: State) -> State:
    step = state.get("current_step", 0)
    plan = state.get("plan", [])
    if step >= len(plan):
        # No more steps: aggregator time
        state["chosen_executor"] = "exec_write"  # placeholder; won't be used
        return state

    instruction = plan[step].lower()

    # Route based on keywords (replace with LLM-based router later)
    if "research" in instruction:
        state["chosen_executor"] = "exec_research"
    elif "draft" in instruction or "response" in instruction:
        state["chosen_executor"] = "exec_write"
    elif "risk" in instruction or "limitation" in instruction:
        state["chosen_executor"] = "exec_risk"
    elif "schedule" in instruction or "checklist" in instruction:
        state["chosen_executor"] = "exec_schedule"
    else:
        state["chosen_executor"] = "exec_write"

    return state


def next_node(state: State) -> str:
    """Conditional edge selector (LangGraph expects a string key)."""
    step = state.get("current_step", 0)
    plan = state.get("plan", [])
    if step >= len(plan):
        return "aggregator"
    return state["chosen_executor"]


# -----------------------------
# 4) Executors (specialists)
# Each executor handles one plan step and writes into state["results"]
# -----------------------------
def _record_result(state: State, output: str) -> State:
    step = state["current_step"]
    subtask = state["plan"][step]
    results = dict(state.get("results", {}))
    results[subtask] = output
    return {"results": results}


def exec_research(state: State) -> State:
    step = state["current_step"]
    subtask = state["plan"][step]
    output = (
        f"[RESEARCH OUTPUT]\n"
        f"- Subtask: {subtask}\n"
        f"- Facts (demo): Agents = goal-driven + tool-using loops; MAS = multiple agents coordinating.\n"
        f"- Notes (demo): Replace this with web/RAG/tool calls in a real system.\n"
    )
    return _record_result(state, output)


def exec_write(state: State) -> State:
    step = state["current_step"]
    subtask = state["plan"][step]
    output = (
        f"[WRITING OUTPUT]\n"
        f"- Subtask: {subtask}\n"
        f"- Draft (demo): Agentic AI systems can plan, act, and iterate using tools and memory to achieve goals.\n"
    )
    return _record_result(state, output)


def exec_risk(state: State) -> State:
    step = state["current_step"]
    subtask = state["plan"][step]
    output = (
        f"[RISK OUTPUT]\n"
        f"- Subtask: {subtask}\n"
        f"- Risks (demo): hallucinated tool use; looping; cost/latency; unsafe actions; missing guardrails.\n"
        f"- Mitigations (demo): step limits, validators, human-in-the-loop, tool permissions, test harness.\n"
    )
    return _record_result(state, output)


def exec_schedule(state: State) -> State:
    step = state["current_step"]
    subtask = state["plan"][step]
    output = (
        f"[SCHEDULE OUTPUT]\n"
        f"- Subtask: {subtask}\n"
        f"1) Define objective + constraints\n"
        f"2) Gather context (docs/data)\n"
        f"3) Execute steps with tool calls\n"
        f"4) Validate output (critic/judge)\n"
        f"5) Produce final deliverable\n"
    )
    return _record_result(state, output)


# -----------------------------
# 5) Step incrementer (after any executor)
# -----------------------------
def advance_step(state: State) -> State:
    return {"current_step": state["current_step"] + 1}


# -----------------------------
# 6) Aggregator node
# -----------------------------
def aggregator(state: State) -> State:
    task = state["task"]
    results = state.get("results", {})

    # Combine outputs in the original plan order
    ordered = []
    for subtask in state.get("plan", []):
        ordered.append(results.get(subtask, f"[MISSING OUTPUT] {subtask}"))

    final = (
        f"=== FINAL AGGREGATED DELIVERABLE ===\n"
        f"Task: {task}\n\n"
        + "\n".join(ordered)
    )
    return {"final": final}


# -----------------------------
# 7) Build the LangGraph
# -----------------------------
def build_graph():
    g = StateGraph(State)

    g.add_node("planner", planner)
    g.add_node("router", route_to_executor)

    g.add_node("exec_research", exec_research)
    g.add_node("exec_write", exec_write)
    g.add_node("exec_risk", exec_risk)
    g.add_node("exec_schedule", exec_schedule)

    g.add_node("advance_step", advance_step)
    g.add_node("aggregator", aggregator)

    # Entry
    g.set_entry_point("planner")

    # Flow:
    # planner -> router -> executor -> advance_step -> router -> ... -> aggregator -> END
    g.add_edge("planner", "router")

    g.add_conditional_edges(
        "router",
        next_node,
        {
            "exec_research": "exec_research",
            "exec_write": "exec_write",
            "exec_risk": "exec_risk",
            "exec_schedule": "exec_schedule",
            "aggregator": "aggregator",
        },
    )

    # After each executor, advance step, then go back to router
    for exec_node in ["exec_research", "exec_write", "exec_risk", "exec_schedule"]:
        g.add_edge(exec_node, "advance_step")

    g.add_edge("advance_step", "router")

    # Aggregator ends
    g.add_edge("aggregator", END)

    return g.compile()


# -----------------------------
# 8) Run demo
# -----------------------------
if __name__ == "__main__":
    app = build_graph()

    initial_state: State = {
        "task": "Introduce Agentic AI and why Multi-Agent Systems are useful",
    }

    result = app.invoke(initial_state)


    print(result["final"])
```

###How to turn this into a “real” Hierarchical MAS (LLM-powered)

* Replace planner() with an LLM that outputs a structured plan: List[str]

* Replace each exec_* with an LLM agent + tools (RAG, web search, SQL, code execution)

* Keep aggregator() as either:

    * a deterministic formatter (safe), or

    * an LLM “editor” that merges and polishes (higher quality)

##Pattern C - Decentralized “Peer-to-Peer” MAS (No boss)

Agents collaborate directly, often through a shared state (like a shared whiteboard).

When to use

* Teams of agents exploring ideas in parallel

* Great for: brainstorming, architecture options, risk analysis

LangGraph tip: Use a “router” node that chooses which agent goes next (or loops through all agents for N rounds).

image.png

```python
"""
LangGraph Demo: Pattern C — Decentralized “Peer-to-Peer” MAS (No boss)

What this demo shows:
- No central coordinator/planner.
- Agents collaborate through a shared "blackboard" state.
- A lightweight "turn manager" node only rotates turns + checks stop conditions
  (it is NOT deciding content or routing; it just advances the loop safely).

Agents in this demo (no LLM required):
- AnalystAgent: adds analysis bullets + open questions
- BuilderAgent: proposes a structured outline + next actions
- SkepticAgent: identifies risks/assumptions + mitigations

Install (if needed):
    pip install langgraph

Run:
    python p2p_mas_demo.py
"""

from __future__ import annotations
from typing import TypedDict, List, Dict, Literal
from langgraph.graph import StateGraph, END


# -----------------------------
# 1) Shared "blackboard" state
# -----------------------------
AgentName = Literal["analyst", "builder", "skeptic"]

class State(TypedDict, total=False):
    topic: str
    blackboard: Dict[str, List[str]]   # shared memory: categories -> list of contributions
    turn_order: List[AgentName]        # fixed rotation order
    turn_idx: int                      # whose turn is it now (index into turn_order)
    rounds: int                        # how many contributions have happened so far
    max_rounds: int                    # stop after N total contributions
    last_agent: AgentName              # who spoke last
    final: str                         # optional final output


# -----------------------------
# 2) Utility helpers
# -----------------------------
def _append(bb: Dict[str, List[str]], key: str, item: str) -> Dict[str, List[str]]:
    new_bb = {k: list(v) for k, v in (bb or {}).items()}
    new_bb.setdefault(key, [])
    new_bb[key].append(item)
    return new_bb


def _current_agent(state: State) -> AgentName:
    return state["turn_order"][state["turn_idx"] % len(state["turn_order"])]


# -----------------------------
# 3) Agents (peers)
# Each writes to the shared blackboard and does NOT route other agents.
# -----------------------------
def analyst_agent(state: State) -> State:
    topic = state["topic"]
    bb = state.get("blackboard", {})

    bb = _append(bb, "analysis", f"Key idea: Break '{topic}' into concepts, benefits, and examples.")
    bb = _append(bb, "questions", "What is the audience level (exec vs technical)?")
    bb = _append(bb, "questions", "What real constraints exist (time, tools, data, safety)?")

    return {"blackboard": bb, "last_agent": "analyst"}


def builder_agent(state: State) -> State:
    topic = state["topic"]
    bb = state.get("blackboard", {})

    bb = _append(bb, "outline", f"Outline for '{topic}': 1) Definition 2) Why it matters 3) Pattern 4) Example 5) Risks")
    bb = _append(bb, "actions", "Next action: create 1 concrete scenario + map it to steps/tools.")
    bb = _append(bb, "actions", "Next action: draft 5-slide structure.")

    return {"blackboard": bb, "last_agent": "builder"}


def skeptic_agent(state: State) -> State:
    topic = state["topic"]
    bb = state.get("blackboard", {})

    bb = _append(bb, "risks", f"Risk: Over-claiming capabilities for '{topic}' without guardrails or evaluation.")
    bb = _append(bb, "assumptions", "Assumption: Agents have reliable tools and clean data.")
    bb = _append(bb, "mitigations", "Mitigation: add step limits, validation checks, and human-in-the-loop.")

    return {"blackboard": bb, "last_agent": "skeptic"}


# -----------------------------
# 4) Turn Manager (NOT a boss)
# It does not decide content; it simply:
# - selects the next agent based on round-robin
# - increments counters
# - stops after max_rounds
# -----------------------------
def turn_manager(state: State) -> State:
    # Initialize defaults if missing
    state.setdefault("turn_order", ["analyst", "builder", "skeptic"])
    state.setdefault("turn_idx", 0)
    state.setdefault("rounds", 0)
    state.setdefault("max_rounds", 6)
    state.setdefault("blackboard", {})

    return state


def choose_next(state: State) -> str:
    # Stop condition
    if state["rounds"] >= state["max_rounds"]:
        return "synthesizer"

    agent = _current_agent(state)
    return {
        "analyst": "analyst_agent",
        "builder": "builder_agent",
        "skeptic": "skeptic_agent",
    }[agent]


def after_agent(state: State) -> State:
    # Advance the round-robin turn and total rounds
    return {
        "turn_idx": state["turn_idx"] + 1,
        "rounds": state["rounds"] + 1,
    }


# -----------------------------
# 5) Synthesizer (optional)
# Combines blackboard into one final output for the demo.
# In real systems, this could be a formatter or an editor agent.
# -----------------------------
def synthesizer(state: State) -> State:
    topic = state["topic"]
    bb = state.get("blackboard", {})

    def section(title: str, key: str) -> str:
        items = bb.get(key, [])
        if not items:
            return f"{title}:\n- (none)\n"
        return f"{title}:\n" + "\n".join(f"- {x}" for x in items) + "\n"

    final = (
        f"=== P2P MAS BLACKBOARD (No boss) ===\n"
        f"Topic: {topic}\n"
        f"Contributions: {state.get('rounds', 0)}\n\n"
        + section("ANALYSIS", "analysis")
        + section("OPEN QUESTIONS", "questions")
        + section("OUTLINE", "outline")
        + section("NEXT ACTIONS", "actions")
        + section("RISKS", "risks")
        + section("ASSUMPTIONS", "assumptions")
        + section("MITIGATIONS", "mitigations")
    )

    return {"final": final}


# -----------------------------
# 6) Build the LangGraph
# -----------------------------
def build_graph():
    g = StateGraph(State)

    g.add_node("turn_manager", turn_manager)

    g.add_node("analyst_agent", analyst_agent)
    g.add_node("builder_agent", builder_agent)
    g.add_node("skeptic_agent", skeptic_agent)

    g.add_node("after_agent", after_agent)
    g.add_node("synthesizer", synthesizer)

    g.set_entry_point("turn_manager")

    # turn_manager decides ONLY who speaks next (round-robin) or to synthesize/end
    g.add_conditional_edges(
        "turn_manager",
        choose_next,
        {
            "analyst_agent": "analyst_agent",
            "builder_agent": "builder_agent",
            "skeptic_agent": "skeptic_agent",
            "synthesizer": "synthesizer",
        },
    )

    # After any agent, we advance the turn + rounds, then go back to turn_manager
    g.add_edge("analyst_agent", "after_agent")
    g.add_edge("builder_agent", "after_agent")
    g.add_edge("skeptic_agent", "after_agent")

    g.add_edge("after_agent", "turn_manager")

    # Synthesizer ends
    g.add_edge("synthesizer", END)

    return g.compile()


# -----------------------------
# 7) Run demo
# -----------------------------
if __name__ == "__main__":
    app = build_graph()

    initial_state: State = {
        "topic": "Intro to Agentic AI: what agents are and why multi-agent systems matter",
        "max_rounds": 6,  # total contributions across all agents
        "turn_order": ["analyst", "builder", "skeptic"],  # you can reorder peers freely
    }

    result = app.invoke(initial_state)

    print(result["final"])
```

##Notes

* There is no coordinator that assigns tasks.

* Agents contribute independently to a shared blackboard.

* The turn_manager is only a safe loop controller (round-robin + stop condition), not a decision-maker.

* In an LLM version, each *_agent node would call an LLM with:

    * the topic

    * the current blackboard

    * a role prompt (Analyst/Builder/Skeptic)

    * tool access if needed

##Pattern D - Debate / Adversarial MAS (Competitive inside, cooperative outcome)

Two agents argue, one judge picks best answer.

When to use

* You want higher quality + reduced blind spots

* Great for: policy decisions, architecture tradeoffs, risk reviews

LangGraph tip: Keep arguments_pro, arguments_con, verdict in state.

image.png

```python
"""
LangGraph Demo: Pattern D — Debate / Adversarial MAS
(Competitive inside, cooperative outcome)

What this demo shows:
- Pro agent argues FOR a proposal
- Con agent argues AGAINST the same proposal
- Judge agent evaluates both and produces a final decision + rationale
- Optional: a second "revision" round where Pro/Con respond to the judge’s feedback

This demo is LLM-free (deterministic text) so it runs anywhere.
You can later swap each agent body with an LLM call.

Install (if needed):
    pip install langgraph

Run:
    python debate_mas_demo.py
"""

from __future__ import annotations
from typing import TypedDict, Literal
from langgraph.graph import StateGraph, END


# -----------------------------
# 1) Shared state
# -----------------------------
class State(TypedDict, total=False):
    motion: str                   # The debated claim/proposal
    context: str                  # Optional context/constraints
    pro_argument: str
    con_argument: str
    judge_verdict: str
    judge_winner: Literal["pro", "con", "mixed"]
    round: int
    max_rounds: int
    final: str


# -----------------------------
# 2) Pro / Con agents
# -----------------------------
def pro_agent(state: State) -> State:
    motion = state["motion"]
    context = state.get("context", "")
    r = state.get("round", 1)

    if r == 1:
        arg = (
            f"[PRO - Round {r}]\n"
            f"Motion: {motion}\n"
            f"Argument:\n"
            f"- Primary benefit: improves outcomes by enabling faster, more consistent decisions.\n"
            f"- Operational benefit: reduces manual effort and frees clinicians for complex cases.\n"
            f"- Safety angle: if paired with guardrails (validation + audit logs), can reduce human error.\n"
            f"{'- Context: ' + context if context else ''}\n"
        )
    else:
        # Respond to judge feedback (simple demo)
        arg = (
            f"[PRO - Round {r}]\n"
            f"Rebuttal / Revision:\n"
            f"- Strengthen governance: require human-in-the-loop for high-risk decisions.\n"
            f"- Add monitoring: drift detection + performance dashboards.\n"
            f"- Clarify scope: decision support, not autonomous diagnosis.\n"
        )

    return {"pro_argument": arg}


def con_agent(state: State) -> State:
    motion = state["motion"]
    context = state.get("context", "")
    r = state.get("round", 1)

    if r == 1:
        arg = (
            f"[CON - Round {r}]\n"
            f"Motion: {motion}\n"
            f"Argument:\n"
            f"- Safety risk: model errors or bias can harm patients at scale.\n"
            f"- Accountability risk: unclear liability if AI advice is wrong.\n"
            f"- Operational risk: over-reliance may reduce clinician vigilance.\n"
            f"- Data risk: privacy/security exposure, especially with sensitive health data.\n"
            f"{'- Context: ' + context if context else ''}\n"
        )
    else:
        # Respond to judge feedback (simple demo)
        arg = (
            f"[CON - Round {r}]\n"
            f"Rebuttal / Revision:\n"
            f"- Even with HITL, alert fatigue and workflow issues can still cause harm.\n"
            f"- Monitoring helps, but rare edge-cases remain hard to detect.\n"
            f"- Recommend phased rollout + strict clinical validation before broad adoption.\n"
        )

    return {"con_argument": arg}


# -----------------------------
# 3) Judge agent
# -----------------------------
def judge_agent(state: State) -> State:
    pro = state.get("pro_argument", "")
    con = state.get("con_argument", "")
    r = state.get("round", 1)

    # Simple scoring heuristics (demo). Swap with LLM judge later.
    pro_score = 0
    con_score = 0

    # Very rough "signals"
    for kw in ["guardrail", "validation", "audit", "human-in-the-loop", "monitor", "scope"]:
        if kw in pro.lower():
            pro_score += 1
    for kw in ["bias", "liability", "privacy", "security", "harm", "over-reliance", "workflow", "fatigue"]:
        if kw in con.lower():
            con_score += 1

    # Decide winner
    if abs(pro_score - con_score) <= 1:
        winner: Literal["pro", "con", "mixed"] = "mixed"
    elif pro_score > con_score:
        winner = "pro"
    else:
        winner = "con"

    verdict = (
        f"[JUDGE - Round {r}]\n"
        f"Assessment:\n"
        f"- Pro score: {pro_score}\n"
        f"- Con score: {con_score}\n"
        f"Winner: {winner.upper()}\n\n"
        f"Verdict:\n"
        f"- Best points (Pro): operational benefits are real IF governance is strong.\n"
        f"- Best points (Con): safety, privacy, and accountability are the main blockers.\n"
        f"- Recommendation: proceed only as *clinical decision support* with HITL, validation, "
        f"monitoring, audit logs, and phased rollout.\n"
    )

    return {"judge_verdict": verdict, "judge_winner": winner}


# -----------------------------
# 4) Optional: Decide whether to do another round
# -----------------------------
def init_round(state: State) -> State:
    state.setdefault("round", 1)
    state.setdefault("max_rounds", 1)
    return state


def next_after_judge(state: State) -> str:
    r = state.get("round", 1)
    max_r = state.get("max_rounds", 1)

    # If more rounds allowed, increment and run another debate round
    if r < max_r:
        return "bump_round"
    return "finalize"


def bump_round(state: State) -> State:
    return {"round": state.get("round", 1) + 1}


# -----------------------------
# 5) Finalize
# -----------------------------
def finalize(state: State) -> State:
    final = (
        "=== DEBATE / ADVERSARIAL MAS RESULT ===\n\n"
        f"Motion:\n{state['motion']}\n\n"
        f"{state.get('pro_argument','')}\n"
        f"{state.get('con_argument','')}\n"
        f"{state.get('judge_verdict','')}\n"
        f"Final Decision: {state.get('judge_winner','mixed').upper()}\n"
    )
    return {"final": final}


# -----------------------------
# 6) Build graph
# -----------------------------
def build_graph():
    g = StateGraph(State)

    g.add_node("init_round", init_round)
    g.add_node("pro_agent", pro_agent)
    g.add_node("con_agent", con_agent)
    g.add_node("judge_agent", judge_agent)
    g.add_node("bump_round", bump_round)
    g.add_node("finalize", finalize)

    g.set_entry_point("init_round")

    # Debate flow: init -> pro -> con -> judge -> (maybe another round) -> finalize -> END
    g.add_edge("init_round", "pro_agent")
    g.add_edge("pro_agent", "con_agent")
    g.add_edge("con_agent", "judge_agent")

    g.add_conditional_edges(
        "judge_agent",
        next_after_judge,
        {
            "bump_round": "bump_round",
            "finalize": "finalize",
        },
    )

    # If bump round, go back to pro/con again
    g.add_edge("bump_round", "pro_agent")

    g.add_edge("finalize", END)

    return g.compile()


# -----------------------------
# 7) Run demo
# -----------------------------
if __name__ == "__main__":
    app = build_graph()

    # Example motion (you can change this)
    initial_state: State = {
        "motion": "Hospitals should adopt AI agents to assist clinicians in triage decisions.",
        "context": "Assume the system is deployed in an emergency department with high patient volume.",
        "max_rounds": 2,  # set to 1 for single-round debate; 2 shows revision after judge feedback
    }

    result = app.invoke(initial_state)

    print(result["final"])
```

###How to upgrade this demo into a real LLM-powered adversarial MAS?

* Replace pro_agent, con_agent, and judge_agent bodies with LLM calls (each with a different system prompt).

* Add tool constraints (e.g., RAG + citations) so arguments must reference evidence.

* Keep the judge deterministic for safety (or use an LLM judge + a final deterministic validator).
