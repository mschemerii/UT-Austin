```
import os
from openai import OpenAI

os.environ["OPENAI_API_KEY"] = "YOUR_KEY_HERE"

client = OpenAI()
MODEL = "gpt-4.1-mini"
```

```
def ask(prompt):
    response = client.responses.create(
        model=MODEL,
        input=prompt
    )
    return response.output_text

print(ask("Say hello in one sentence"))
```

## Zero-Shot Prompting

```

prompt = '''
Classify the sentiment of this review as Positive, Negative, or Neutral.

Review: "The battery life is decent but the camera is disappointing."
'''
print(ask(prompt))
```

## Few-Shot Prompting

```

prompt = '''
Classify sentiment.

Review: "Amazing display and speed."
Sentiment: Positive

Review: "Stopped working after two days."
Sentiment: Negative

Review: "It's okay, nothing special."
Sentiment:
'''
print(ask(prompt))
```

## Self-Consistency

```

from collections import Counter

question = "If a train travels 60 miles in 1 hour, how far in 3 hours?"

answers = [ask(question) for _ in range(5)]
print(answers)

majority = Counter(answers).most_common(1)[0]
print("\nMajority Answer:", majority)
```

## Generated Knowledge Prompting

```

prompt = '''
Before answering, list key facts about why exercise improves mental health.
Then give a final answer in 2 sentences.
'''
print(ask(prompt))
```

## Metacognitive Prompting

```

prompt = '''
Solve the problem and then verify your answer.

Problem:
A shop gives 20% discount on $50. What is the final price?
'''
print(ask(prompt))
```

## Chain-of-Thought

```

prompt = '''
Solve step-by-step:

A restaurant had 45 customers.
15 left and 20 more arrived.
How many customers now?
'''
print(ask(prompt))
```

## Automatic Chain-of-Thought

```

prompt = '''
Let's think step by step.

If each server handles 120 requests/min,
how many can 5 servers handle?
'''
print(ask(prompt))
```

## Tree of Thoughts

```

prompt = '''
Generate 3 strategies to reduce warehouse shipping cost.
Then evaluate each briefly.
Finally choose the best one.
'''
print(ask(prompt))
```

## Graph of Thoughts

```

prompt = '''
Break down launching an ML platform into a dependency graph.
Return as bullet hierarchy.
'''
print(ask(prompt))
```

## RAG Style Prompt

```

context = '''
Return policy:
Items can be returned within 30 days with receipt.
'''

question = "Can I return an item after 2 weeks?"

prompt = f'''
Use ONLY the context below.

Context:
{context}

Question:
{question}
'''
print(ask(prompt))
```

## ReAct Prompting

```

prompt = '''
You are an analyst.

First reason about the problem,
then give the final answer.

Problem:
Should a company reduce prices or increase marketing if sales drop?
'''
print(ask(prompt))
```

## Algorithm of Thoughts

```

prompt = '''
Use an algorithmic approach:

1. Identify variables
2. Evaluate options
3. Choose optimal solution

Problem:
Select the best cloud instance for heavy GPU inference with cost constraints.
'''
print(ask(prompt))
```

## Automatic Prompt Engineer (APE)

```

prompt = '''
Improve this prompt for clarity and performance:

'Explain demand forecasting.'
'''
print(ask(prompt))
```
